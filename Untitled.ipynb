```python
import instaloader

L = instaloader.Instaloader()

# Load the active session from your browser cookie
try:
    print("Attempting to import saved session...")
    # The correct method name is load_session_from_file
    L.load_session_from_file("khattubanisa")
    print("Saved session loaded successfully!")
except FileNotFoundError:
    print("No saved session found. Importing directly from your active browser...")
    import browser_cookie3
    # Update the session cookies using your active Chrome session
    L.context._session.cookies.update(browser_cookie3.chrome(domain_name='.instagram.com'))
    
    # Save the session to a file so it works next time without browser_cookie3
    L.save_session_to_file()
    print("Session imported from browser and saved successfully!")

# Now run your scraper code safely
try:
    profile_name = "sakshilalwani_nutritionist"
    profile = instaloader.Profile.from_username(L.context, profile_name)
    
    for post in profile.get_posts():
        print(f"Downloading post from date: {post.date}")
        L.download_post(post, target=profile_name)
        break # Testing with just one post first to avoid rate-limits
        
except Exception as e:
    print(f"An error occurred during scraping: {e}")
```

```python
import instaloader
import json

L = instaloader.Instaloader()

# Load session from file (you need to have created this earlier)
USERNAME = "sakshilalwani_nutritionist"
try:
    L.load_session_from_file(USERNAME)
    print("✅ Session loaded successfully!")
except:
    print("❌ No session file found. Please login first using Solution 1.")
    exit()

# Now scrape the post
shortcode = "BRS2Q9IjznM"
post = instaloader.Post.from_shortcode(L.context, shortcode)

comments = []
for comment in post.get_comments():
    comments.append({
        "username": comment.owner.username,
        "text": comment.text,
        "created_at": str(comment.created_at),
        "likes": comment.likes_count
    })

print(f"Scraped {len(comments)} comments")
```
